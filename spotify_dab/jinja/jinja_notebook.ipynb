```python
from jinja2 import Template
```

```python
from pyspark.sql.functions import *

parameters = [
    {
        "table" : "spotify_cata.silver.factstream",
        "alias" : "factstream",
        "cols" : "factstream.stream_id,factstream.listen_duration"
    },
    {
        "table" : "spotify_cata.silver.dimuser",
        "alias" : "dimuser",
        "cols" : "dimuser.user_id,dimuser.user_name",
        "condition" : "dimuser.user_id = factstream.user_id"
    },
    {
        "table" : "spotify_cata.silver.dimtrack",
        "alias" : "dimtrack",
        "cols" : "dimtrack.track_id,dimtrack.track_name",
        "condition" : "dimtrack.track_id = factstream.track_id"
    }
]
```

```python

# select stream_id, listen_duration from factstream left join dimuser on factstream.user_id = dimuser.user_id left join dimtrack on dimtrack.track_id = factstream.track_id

query_text = """
    
    
        SELECT
            {% for params in parameters %}
                {{ params.cols }}
                {% if not loop.last %}
                    ,
                {% endif %}
            {% endfor %}
        FROM 
            {{ parameters[0].table }} AS {{ parameters[0].alias }}

            {% for p in parameters[1:] %}
            
            LEFT JOIN {{ p.table }} AS {{ p.alias }}
            ON {{ p.condition }}
            
            {% endfor %}
"""
```

```python
jinja_sql_str = Template(query_text)
query = jinja_sql_str.render(parameters=parameters)
print(query)
```
